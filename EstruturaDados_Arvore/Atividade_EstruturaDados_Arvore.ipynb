```python
class No:
    def __init__(self, chave):
        self.chave = chave
        self.esquerda = None
        self.direita = None
        self.p = None  # Ponteiro para o pai (parent)


class ArvoreBuscaBinaria:
    def __init__(self):
        self.raiz = None

    # Tradução exata de TREE-INSERT(T, z)
    def tree_insert(self, z): #TREE-INSERT: é a inserção de um novo nó
        y = None
        x = self.raiz

        # 1. Percorre a árvore até encontrar a folha/posição NIL
        while x is not None:
            y = x
            if z.chave < x.chave:
                x = x.esquerda
            else:
                x = x.direita

        # 2. Define o pai do novo nó
        z.p = y

        # 3. Conecta o novo nó ao nó pai 'y' (ou torna-se a raiz)
        if y is None:
            self.raiz = z  # A árvore T estava vazia
        elif z.chave < y.chave:
            y.esquerda = z
        else:
            y.direita = z

    # Tradução exata de INORDER-TREE-WALK(x)
    def inorder_walk(self, x):
        if x is not None:
            self.inorder_walk(x.esquerda)
            print(x.chave, end=" ")
            self.inorder_walk(x.direita)


# --- Testando a execução ---
arvore = ArvoreBuscaBinaria()

# Inserindo exatamente os valores descritos no comentário
valores = [15, 6, 18, 3, 7, 13, 9]

for val in valores:
    novo_no = No(val)
    arvore.tree_insert(novo_no)

print("Conteúdo da árvore impresso em ordem (In-Order):")
arvore.inorder_walk(arvore.raiz)
print()  # Quebra de linha final

# Saída: 3 6 7 9 13 15 18
```
